```python
import pandas as pd
import regex as re
import os
from pathlib import Path
```

```python
text = Path('./ghayra_mujdin.txt').read_text('utf-8').splitlines()
text = [line.strip() for line in text]

df = pd.DataFrame(text).rename(columns={0:'VERSES'})
df['IDX'] = df.index
df['V_ID'] = df['IDX'] // 2
df['GRP'] = df.apply(lambda x: tuple(
    [x['IDX'],  x['VERSES']]
) , axis=1)

df = df.groupby(['V_ID'], as_index=False).agg(
    {'GRP': lambda x: list(x)}
).sort_values(['V_ID']).reset_index(drop=True)
df['GRP'] = df['GRP'].apply(lambda x: sorted(x, key=lambda y: y[0]))
df['VERSES'] = df['GRP'].apply(lambda x: tuple([i[1] for i in x]))
df
```

```python
df['VERSES'].to_json('./ghayra_mujdin.json', orient='records', force_ascii=False)
```
