```python
import pandas as pd
import regex as re
import os
from pathlib import Path
import json
from scrapy import Selector
```

```python
text = Path('./text.txt').read_text('utf-8')
text
```

```python
l = Selector(text = str(text)).xpath('//h3').getall()
```

```python
r = [Selector(text = str(i)).xpath('//text()').getall() for i in l]
r = [' '.join(i) for i in r]
r = [re.sub(r'\s+', ' ', i) for i in r]
r
```
